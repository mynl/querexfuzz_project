```python
# separate startup from here
%load_ext autotime
%load_ext IPython.extensions.autoreload
%autoreload 2
import logging 
import yaml 
from pathlib import Path

def setup_logging(config_path):
    with config_path.open("r") as f:
        config = yaml.safe_load(f)
        logging.config.dictConfig(config)

setup_logging(Path("logconfig.yaml"))

from skimmatch import FuzzyMatcherMultiHi
from greater_tables.core import GT
from greater_tables.fabrications import quick_fab
from querexfuzz.config import FuzzyConfig
from querexfuzz import Querexfuzz, querexfuzz_from_df
import querexfuzz.dates as qfd
from functools import partial
fGT = partial(GT,  default_date_str="%Y-%m-%d %H:%M:%S.%f", 
             max_table_inch_width=12, table_font_pt_size=9,
             tikz=False)
```

```python
# a larger dataframe
col_names = ['name', 'bio', 'size', 'number', 'accessed', 'create', 'mod', 'omit']
testdf = quick_fab(rows=10000, data_spec='ss20fidddi', 
                   metric_name_spec=col_names
                  )
fGT(testdf.head(2))
```

```python
# default and helper 
qq = querexfuzz_from_df(testdf, bang_field='name', recent_field='mod', fuzzy_fields=['bio'])
qq.config
```

```python
fGT(testdf.q('top 10 select name, bio, number, -create, -mod, -accessed # waist'))
```

```python
fGT(testdf.q('recent top 10 select name, bio, number, -create, -mod, -accessed # waist'))
```

```python
fGT(testdf.q('order by -create top 3 select -bio, -accessed, -mod'), default_date_str="%Y-%m-%d %H:%M:%S.%f")
```
